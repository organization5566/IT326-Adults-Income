```python

# ==========================================================
# Advanced Preprocessing (Person 4): Feature Selection + Discretization
# ==========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

from sklearn.feature_selection import SelectKBest, chi2
from sklearn.preprocessing import KBinsDiscretizer

# ---------------------------
# 0) Load / prepare base data
# ---------------------------
# Try to reuse existing df if available; otherwise load from Dataset/Raw_dataset.csv
try:
    df
except NameError:
    csv_path = Path("Dataset/Raw_dataset.csv")
    if not csv_path.exists():
        csv_path = Path("./Raw_dataset.csv")
    df = pd.read_csv(csv_path)

# Normalize income labels and basic cleaning for safety
if 'income' in df.columns:
    df['income'] = df['income'].astype(str).str.strip().str.replace('.', '', regex=False)
    df['income'] = df['income'].replace({'>50K':1, '<=50K':0, '>50K,':1, '<=50K,':0})
    # If it's still not numeric, try factorize with ensuring 1 is ">50K"
    if not pd.api.types.is_numeric_dtype(df['income']):
        # Heuristic: set highest class as 1
        df['income'] = (df['income'] == '>50K').astype(int)

# Replace '?' with NaN then drop rows with NaN to keep chi2 happy
df = df.replace('?', np.nan).dropna(axis=0)

# If an encoded frame exists, reuse it; otherwise create one
try:
    df_encoded
    # ensure income is present and numeric
    if not pd.api.types.is_numeric_dtype(df_encoded['income']):
        df_encoded['income'] = (df_encoded['income'].astype(str).str.strip().str.replace('.', '', regex=False) == '>50K').astype(int)
except NameError:
    # One-hot encode categoricals; keep numeric as is
    # Ensure 'income' remains as target
    feature_cols = [c for c in df.columns if c != 'income']
    df_features = pd.get_dummies(df[feature_cols], drop_first=True)
    # Map income to 0/1 if needed
    y_series = df['income']
    if not pd.api.types.is_numeric_dtype(y_series):
        y_series = (y_series.astype(str).str.strip().str.replace('.', '', regex=False) == '>50K').astype(int)
    df_encoded = pd.concat([df_features, y_series.rename('income')], axis=1)

# Guard: remove any potential negative values for chi2 (shouldn't exist after one-hot, but safe)
X_all = df_encoded.drop(columns=['income'], errors='ignore')
X_all = X_all.clip(lower=0)
y = df_encoded['income']

# ---------------------------
# 1) Feature Selection (SelectKBest + chi2)
# ---------------------------
k = 10 if X_all.shape[1] >= 10 else X_all.shape[1]
selector = SelectKBest(score_func=chi2, k=k)
X_selected = selector.fit_transform(X_all, y)
mask = selector.get_support()

selected_features = X_all.columns[mask].tolist()
scores = selector.scores_[mask]

feature_scores = pd.DataFrame({'Feature': selected_features, 'Score': scores}).sort_values('Score', ascending=False)
print("Top selected features (by chi2):")
print(feature_scores.to_string(index=False))

# Plot (matplotlib only; no specific colors)
plt.figure(figsize=(8, 4))
plt.barh(feature_scores['Feature'], feature_scores['Score'])
plt.title("Top Selected Features (Chi2)")
plt.xlabel("Chi2 score")
plt.ylabel("Feature")
plt.gca().invert_yaxis()
plt.show()

# Build working frame with selected features + target
df_selected = pd.concat([df_encoded[selected_features], y.rename('income')], axis=1)

# ---------------------------
# 2) Discretization (KBinsDiscretizer) for numeric columns
# ---------------------------
disc_cols = []
for cand in ['age', 'hours-per-week', 'capital-gain', 'capital-loss']:
    if cand in df.columns and cand in df_selected.columns:  # after selection
        disc_cols.append(cand)

if disc_cols:
    discretizer = KBinsDiscretizer(n_bins=4, encode='ordinal', strategy='uniform')
    binned = discretizer.fit_transform(df_selected[disc_cols])
    for i, col in enumerate(disc_cols):
        df_selected[col + "_binned"] = binned[:, i].astype(int)
    print(f"Discretized columns: {disc_cols}")
else:
    print("No numeric columns from the original set survived selection for discretization; skipping discretization step.")

# ---------------------------
# 3) Save preprocessed dataset
# ---------------------------
out_path = Path("Dataset/Preprocessed_dataset.csv")
out_path.parent.mkdir(parents=True, exist_ok=True)
df_selected.to_csv(out_path, index=False)
print(f"✅ Saved: {out_path}  (rows={df_selected.shape[0]}, cols={df_selected.shape[1]})")

# ---------------------------
# 4) Snapshots: Raw vs. Preprocessed
# ---------------------------
print("\n--- Raw dataset sample ---")
display(df.head(3))
print("\n--- Preprocessed dataset sample ---")
display(df_selected.head(3))
```


## 📄 Documentation — Advanced Preprocessing

**Feature Selection (SelectKBest + Chi-Squared):**  
We applied the Chi-Squared test to rank features by their association with the binary target `income`.  
Selecting the top features reduces dimensionality, removes irrelevant variables, and helps downstream models (Decision Trees / K-Means) focus on informative signals.

**Discretization (KBinsDiscretizer):**  
We discretized selected numeric attributes (e.g., `age`, `hours-per-week`, `capital-gain`, `capital-loss`) into 4 ordinal bins.  
This improves interpretability, reduces the impact of extreme values, and can make distance-based methods (like K-Means) more stable in later phases.

**How these steps improved data quality:**  
- Lower noise and redundancy → leaner feature space and faster training.  
- More robust handling of skewed numeric distributions via binning.  
- A clearer structure that benefits both **classification** and **clustering** in Phase 3.

**Snapshots:**  
- *Raw Dataset:* Mixed types, potential missing markers (`?`), and continuous ranges.  
- *Preprocessed Dataset:* Top-ranked features retained + added `_binned` features where applicable.  
This alignment sets up the dataset for reproducible modeling and fair comparisons in the next phase.

**Data Cleaning**
Missing values
in this step missing values were replaced using mode imputation, Duplicates were removed .Outliers were handled using the IQR method.

```python
df = df.replace("?", pd.NA)
```

```python
cols_with_missing = ["workclass", "occupation", "native-country"]
for col in cols_with_missing:
  mode_value = df[col].mode()[0]
  df[col] = df[col].fillna(mode_value)
```

The colmuns *woekclass* , *occupation* and *native-country* contanied missing values, these were replaced by most frequent(mode) values in each column.

**Label Encoding - Gender**


```python
le = LabelEncoder()
df["gender"] = le.fit_transform(df["gender"])
```

The gender column was converted to numeric from using Label Encoding so algorthims can interpet it correctly(Male=1 , Female=0).

**One-Hot Encoding - Multi-category coulmns**



```python
df = pd.get_dummies(df, columns=["workclass", "education", "marital-status"] ,drop_first=True)
```

Multi-class categorial feauters were transformed using One-Hot Encoding to prevent any false ordering between categories to creat binary indicator columns.

**Normalization - Min-Max Scaling**

```python
numeric_cols = ["age", "hours-per-week", "capital-gain", "capital-loss"]
scaler = MinMaxScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])
```

Numerical columns were scaled to the range [0, 1] using Min-Max Normalization
